```python
asian_cars = df[df['origin'] == 'Asia']
asian_cars
```

```python
ac_vh_my = asian_cars[['vehicle_weight', 'model_year']]
ac_vh_my
```

```python
ac_vh_my_f7 = ac_vh_my.head(7)
ac_vh_my_f7
```

```python
X = ac_vh_my_f7.to_numpy()
X
```

```python
XTX = X.T @ X
XTX
```

```python
XTX_inv = np.linalg.pinv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = XTX_inv.dot(X.T).dot(y)
w
```
